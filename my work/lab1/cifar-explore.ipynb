```python
def extract_image(batch, index):
    img = batch[index*3072: index*3072+3072]
    img_red = img[0:1042]
    img_green = img[1042:2084]
    img_blue = img[2084:]
    return (img_red, img_green, img_blue)
```

```python
from pathlib import Path
import numpy as np

dir_path = Path('data/cifar-10-batches-py')

for item in dir_path.iterdir():
    if item.is_file() and "batch" in item.name:
        print(f"File: {item.name}: {item,stat().st_size}")
        batch = np.fromfile(str(item.path))
        first = extract_image(batch, 0)
        plt.imshow(np.reshape(np.concatenate(first), (32, 32, 3)))
```
